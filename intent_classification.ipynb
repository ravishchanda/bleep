#import necessary libraries

```python
from transformers import pipeline
```

```python
from transformers import AutoModelForSequenceClassification, TextClassificationPipeline,AutoTokenizer
```

```python
model_name = 'qanastek/XLMRoberta-Alexa-Intents-Classification'
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
model = AutoModelForSequenceClassification.from_pretrained(model_name)
```

```python
classifier = TextClassificationPipeline(model=model,tokenizer=tokenizer)
```

```python
res = classifier("what is the weather today in Miami ?")
```

```python
res[0]['label']
```

```python
!pip install gradio
```

```python
import gradio as gr
```

```python
def intent_classifier(text):
    res=classifier(text)
    return res[0]['label']
```

```python
iface = gr.Interface(
    fn = intent_classifier,
    inputs = gr.Textbox(lines=2,placeholder = "Enter your query.."),
    outputs = "text",
    title = "Intent classification using HuggingFace"
)
```

```python
iface.launch()
```

#Named Entity recognization

```python
from transformers import AutoTokenizer, AutoModelForTokenClassification
```

```python
bert_model_name = 'dslim/bert-base-NER'
```

```python
b_tokenizer = AutoTokenizer.from_pretrained(bert_model_name)
```

```python
b_model = AutoModelForTokenClassification.from_pretrained(bert_model_name)
```

```python
b_nlp = pipeline("ner",model=b_model,tokenizer = b_tokenizer)
```

```python
def bert_classifier(text):
    res=b_nlp(text)
    return res[0]['entity']
```

```python
b_iface = gr.Interface(
    fn = bert_classifier,
    inputs = gr.Textbox(lines=2,placeholder = "Enter your query.."),
    outputs = "text",
    title = "NER using HuggingFace"
)
```

```python
b_iface.launch()
```
